# Python Programming Basics

A practical introduction to core Python concepts. Run the code cells in order in Jupyter.

## 1. Values, variables, and comments
Python is dynamically typed. Use descriptive names in `snake_case`.

```python
# A comment starts with #
name = "Ada"
age = 36
height = 1.68
is_learning = True
print(name, age, height, is_learning)
print(type(age))
```

## 2. Built-in data types
Common types include integers, floats, strings, booleans, `None`, and collection types.

```python
whole_number = 7
decimal_number = 3.14
message = 'Hello, Python!'
enabled = False
not_set = None
print(type(whole_number), type(decimal_number), type(message), type(enabled), type(not_set))
print(isinstance(whole_number, int))
```

## 3. Operators and conversions
Use `==` to compare values and `=` to assign. Convert types explicitly when needed.

```python
a, b = 10, 3
print(a + b, a - b, a * b, a / b)
print(a // b, a % b, a ** b)
print(a > b, a == b, a != b)
print(True and False, True or False, not True)
number_text = '42'
number = int(number_text)
print(number, str(number), float(number))
```

## 4. Strings
Strings are immutable sequences of characters. Use f-strings for readable formatting.

```python
first_name = 'Grace'
greeting = f'Hello, {first_name}!'
print(greeting)
print(greeting.upper(), greeting.lower())
print('  Python  '.strip())
words = 'red,green,blue'.split(',')
print(words, ' / '.join(words))
print(first_name[0], first_name[-1], first_name[1:4])
```

## 5. Collections: lists, tuples, sets, and dictionaries

```python
# Lists are ordered and mutable.
colors = ['red', 'green']
colors.append('blue')
colors[0] = 'orange'
print(colors, colors[0], colors[-1], colors[1:])

# Tuples are ordered and immutable.
point = (4, 9)
x, y = point
print(x, y)

# Sets contain unique values.
unique_numbers = {1, 2, 2, 3}
unique_numbers.add(4)
print(unique_numbers)

# Dictionaries map keys to values.
person = {'name': 'Ada', 'age': 36}
person['city'] = 'London'
print(person['name'], person.get('occupation', 'Not specified'))
print(person.keys(), person.values())
```

## 6. Conditions
Indentation defines blocks in Python; use four spaces per indentation level.

```python
score = 82
if score >= 90:
    grade = 'A'
elif score >= 80:
    grade = 'B'
else:
    grade = 'C or below'
print(grade)

# Conditional expression
status = 'pass' if score >= 50 else 'try again'
print(status)
```

## 7. Loops and comprehensions

```python
for color in colors:
    print(color)

for index, color in enumerate(colors, start=1):
    print(index, color)

for number in range(3):
    print(number)

countdown = 3
while countdown > 0:
    print(countdown)
    countdown -= 1

squares = [number ** 2 for number in range(6) if number % 2 == 0]
print(squares)
```

## 8. Functions
Functions package reusable logic. Parameters can have defaults; `*args` accepts extra positional arguments.

```python
def greet(person_name='friend'):
    return f'Hello, {person_name}!'

def average(*numbers):
    if not numbers:
        return 0
    return sum(numbers) / len(numbers)

double = lambda value: value * 2  # small anonymous function
print(greet(), greet('Lin'))
print(average(2, 4, 6), double(5))
```

## 9. Modules and imports
Modules are Python files. Import standard-library or installed packages to reuse functionality.

```python
import math
from random import choice
print(math.sqrt(81))
print(choice(['tea', 'coffee', 'water']))

# In a .py file, this guard runs code only when that file is executed directly:
# if __name__ == '__main__':
#     print('Run this file directly')
```

## 10. Errors and exceptions
Use `try` and `except` to handle expected errors; use `finally` for cleanup that must happen either way.

```python
text = '123'
try:
    value = int(text)
except ValueError as error:
    print(f'Could not convert: {error}')
else:
    print('Converted value:', value)
finally:
    print('Conversion attempt complete')

# Raise an exception for invalid input:
# raise ValueError('A helpful explanation of the problem')
```

## 11. Files and JSON
The `with` statement closes a file automatically. This example writes and reads a small local file.

```python
from pathlib import Path
import json

file_path = Path('example.txt')
file_path.write_text('Hello from Python!\n', encoding='utf-8')
print(file_path.read_text(encoding='utf-8').strip())

data = {'language': 'Python', 'version': 3}
json_text = json.dumps(data, indent=2)
print(json_text)
print(json.loads(json_text))
```

## 12. Classes and objects
A class defines a type of object. `__init__` initializes each instance, and `self` refers to that instance.

```python
class Dog:
    kind = 'canine'  # class attribute shared by instances

    def __init__(self, name):
        self.name = name

    def speak(self):
        return f'{self.name} says woof!'

my_dog = Dog('Rex')
print(my_dog.name, my_dog.kind, my_dog.speak())
```

## 13. Iterators and generators
Generators yield values one at a time, which can be useful for processing sequences lazily.

```python
def count_up_to(limit):
    number = 1
    while number <= limit:
        yield number
        number += 1

print(list(count_up_to(5)))
```

## 14. Useful built-ins and good habits
Use `help(value)` and `dir(value)` to explore objects. Prefer clear names, small functions, and readable code.

```python
values = [4, 1, 7, 2]
print(len(values), min(values), max(values), sum(values), sorted(values))
print(any(value > 5 for value in values))
print(all(value > 0 for value in values))

# Common reminders:
# - Python uses indentation to group statements.
# - Lists and dictionaries are mutable; strings and tuples are immutable.
# - Use == for equality and 'is' mainly for identity checks such as value is None.
# - Keep secrets and passwords out of source code.
```
